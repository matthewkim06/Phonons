# Run the phonondft Sim2L

Before running, relax your material in [CellRelaxDFT](https://nanohub.org/tools/cellrelaxdft) and copy that run's `squid_id` below.

There is no cutoff convergence scan. `ecutwfc` defaults to **80 Ry** (a guess for norm-conserving pseudopotentials); edit it below, or set 0 to reuse the cutoff of the relaxation. `ecutrho` = `ecutrho_dual` x `ecutwfc` (default 8).

Metals: set `metal = True` (no Born charges). Insulators get the LO-TO correction from DFPT Born charges; a warning is returned when the DFT gap is small, since the splitting is then underestimated (e.g. GaAs with PBE).

Outputs not computed are stored as -1.

```python
# Import simtool libraries
from simtool import findInstalledSimToolNotebooks,searchForSimTool
from simtool import getSimToolInputs,getSimToolOutputs,Run
```

```python
simToolName = "phonondft"
simToolLocation = searchForSimTool(simToolName)
for key in simToolLocation.keys():
    print("%18s = %s" % (key,simToolLocation[key]))
```

```python
installedSimToolNotebooks = findInstalledSimToolNotebooks(simToolName,returnString=True)
print(installedSimToolNotebooks)
```

```python
# get the list of inputs
inputs = getSimToolInputs(simToolLocation)
print(inputs)
```

```python
# get the list of outputs
outputs = getSimToolOutputs(simToolLocation)
print(outputs)
```

```python
# Enter the squid of your finished cellrelaxdft run (shown in that run's results),
# e.g. "cellrelaxdft/r111/0123abcd..."
squid_id = ""

assert squid_id.strip(), "Paste your cellrelaxdft squid_id above before running"
```

```python
inputs['squid_id'].value = squid_id
inputs['ecutwfc'].value = 80          # guessed cutoff (Ry): edit for your pseudopotentials
inputs['metal'].value = False

# Optional settings (defaults shown)
# inputs['ecutrho_dual'].value = 8
# inputs['kspacing'].value = 0.25            # 1/Angstrom
# inputs['supercell_length'].value = 10.0    # Angstrom
# inputs['compute_born'].value = True
# inputs['plot_dos'].value = True
# inputs['ncpus'].value = 16
# inputs['walltime'].value = 120             # minutes per QE job
# inputs['max_parallel_submits'].value = 8
```

```python
inputs
```

```python
r = Run(simToolLocation,inputs)
```

```python
r.getResultSummary()
```

```python
# Full outputs; an array of -1 means "not computed"
import numpy as np
np.set_printoptions(precision=3, suppress=True, linewidth=120)
for name in ["spacegroup", "cutoff_note", "supercell", "gamma_phonon_frequencies",
             "min_frequency", "eps_inf", "born_charges", "band_gap", "warnings"]:
    print(f"{name}:\n{r.read(name)}\n")
```
